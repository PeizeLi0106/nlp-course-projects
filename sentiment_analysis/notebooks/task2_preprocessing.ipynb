```python
# Import Libraries
import pandas as pd
import re
import nltk
from nltk.stem import WordNetLemmatizer

# Load the Twitter sentiment dataset
df = pd.read_csv("../data/TweetSentiment.csv", encoding="latin1")
# Shows where empty values exist
print(df.isna().sum())
# Drops the empty value
df = df.dropna(subset=['text'])
print(df['text'].isna().sum())
df = df[['text', 'sentiment']]
```

```python
# Converts the texts to lowercased
df["text"] = df["text"].str.lower()
```

```python
# Removes URLs from the texts
df['text'] = df['text'].str.replace(
    r'http\S+|www\.\S+',
    '',
    regex=True
)
```

```python
# Removes all characters that are not letters, numbers, underscores, whitespaces, and asterisks
df['text'] = df['text'].str.replace(
    r"[^\w\s*]",
    '',
    regex=True
)
```

```python
# Download resources needed
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

# Function to lemmatize a single tweet text
def lemmatize_text(text):
    if not isinstance(text, str):
        return ""
    words = text.split() # tokenization
    lemmatized_words = []

    for word in words:
        lemma = lemmatizer.lemmatize(word)
        lemmatized_words.append(lemma)

    return ' '.join(lemmatized_words)

# Apply lemmatization to all tweet texts
df['text'] = df['text'].apply(lemmatize_text)


```

```python
# Drop rows where texts became empty after preprocessing
df = df[df['text'].str.strip() != '']
df.head()
```

```python
# download the preprocessed TweetSentiment dataset to the local working directory and use it for the following tasks
df.to_csv(
    "../data/TweetSentiment_preprocessed.csv",
    index=False
)
```
